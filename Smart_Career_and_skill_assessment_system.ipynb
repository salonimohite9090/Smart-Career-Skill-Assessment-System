```python
# Smart Career and Skill Assessment System


# ==========================================
# PARENT CLASS
# ==========================================

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age


# ==========================================
# CHILD CLASS
# INHERITANCE
# ==========================================

class Student(Person):

    def __init__(self, name, age, education):
        super().__init__(name, age)
        self.education = education

    def display_student(self):

        print("Name       :", self.name)
        print("Age        :", self.age)
        print("Education  :", self.education)


# ==========================================
# SKILL ASSESSMENT CLASS
# ==========================================

class SkillAssessment:

    def __init__(self):
        self.skills = {}

    def get_rating(self, skill):

        while True:

            try:
                rating = int(input(f"Rate your {skill} skill (1-10): "))

                if 1 <= rating <= 10:
                    return rating

                else:
                    print("Please enter a rating between 1 and 10.")

            except ValueError:
                print("Please enter a valid number.")

    def take_assessment(self):

        print("\n========== SKILL ASSESSMENT ==========")
        print("Give rating from 1 to 10.")

        self.skills["Python"] = self.get_rating("Python")
        self.skills["SQL"] = self.get_rating("SQL")
        self.skills["Communication"] = self.get_rating("Communication")
        self.skills["Problem Solving"] = self.get_rating("Problem Solving")
        self.skills["Excel"] = self.get_rating("Excel")
        self.skills["Web Development"] = self.get_rating("Web Development")

        return self.skills


# ==========================================
# CAREER ANALYZER CLASS
# ==========================================

class CareerAnalyzer:

    def __init__(self, skills):

        self.skills = skills
        self.career_scores = {}

    def assess_career(self):

        # Career scores

        data_analyst = (
            self.skills["Python"] +
            self.skills["SQL"] +
            self.skills["Excel"] +
            self.skills["Problem Solving"]
        )

        python_developer = (
            self.skills["Python"] +
            self.skills["Problem Solving"] +
            self.skills["Communication"]
        )

        web_developer = (
            self.skills["Web Development"] +
            self.skills["Python"] +
            self.skills["Problem Solving"]
        )

        business_analyst = (
            self.skills["SQL"] +
            self.skills["Excel"] +
            self.skills["Communication"] +
            self.skills["Problem Solving"]
        )

        software_developer = (
            self.skills["Python"] +
            self.skills["Problem Solving"] +
            self.skills["Web Development"]
        )

        self.career_scores = {
            "Data Analyst": data_analyst,
            "Python Developer": python_developer,
            "Web Developer": web_developer,
            "Business Analyst": business_analyst,
            "Software Developer": software_developer
        }

        # Find highest score

        highest_score = max(self.career_scores.values())

        recommended_careers = []

        for career, score in self.career_scores.items():

            if score == highest_score:
                recommended_careers.append(career)

        return recommended_careers

    def show_skill_analysis(self):

        print("\n========== SKILL ANALYSIS ==========")

        strong_skills = []
        weak_skills = []

        for skill, rating in self.skills.items():

            if rating >= 7:
                strong_skills.append(skill)

            elif rating <= 5:
                weak_skills.append(skill)

        print("\nStrong Skills:")

        if len(strong_skills) > 0:

            for skill in strong_skills:
                print("✓", skill)

        else:
            print("No strong skills found.")

        print("\nSkills to Improve:")

        if len(weak_skills) > 0:

            for skill in weak_skills:
                print("→", skill)

        else:
            print("Good! No major skill gaps.")


# ==========================================
# CAREER ROADMAP CLASS
# ==========================================

class CareerRoadmap:

    def show_roadmap(self, career):

        print("\n========== CAREER ROADMAP ==========")

        if career == "Data Analyst":

            print("1. Learn Advanced SQL")
            print("2. Improve Excel")
            print("3. Learn Power BI")
            print("4. Learn Statistics")
            print("5. Practice Data Analysis")

        elif career == "Python Developer":

            print("1. Improve Python")
            print("2. Learn OOP")
            print("3. Learn Django or Flask")
            print("4. Learn APIs")
            print("5. Build Python Projects")

        elif career == "Web Developer":

            print("1. Learn HTML")
            print("2. Learn CSS")
            print("3. Learn JavaScript")
            print("4. Learn Python Web Framework")
            print("5. Build Websites")

        elif career == "Business Analyst":

            print("1. Improve Communication")
            print("2. Learn Advanced Excel")
            print("3. Learn SQL")
            print("4. Learn Power BI")
            print("5. Learn Business Analysis")

        elif career == "Software Developer":

            print("1. Improve Python")
            print("2. Learn OOP")
            print("3. Practice DSA")
            print("4. Learn Git and GitHub")
            print("5. Build Software Projects")


# ==========================================
# MAIN APPLICATION CLASS
# ==========================================

class CareerApp:

    def __init__(self):

        self.student = None
        self.assessment = SkillAssessment()
        self.analyzer = None
        self.roadmap = CareerRoadmap()

    def get_student_details(self):

        print("\n========== STUDENT DETAILS ==========")

        # Name validation

        while True:

            name = input("Enter your name: ").strip()

            if name:
                break

            print("Name cannot be empty.")

        # Age validation

        while True:

            try:
                age = int(input("Enter your age: "))

                if 1 <= age <= 100:
                    break

                print("Please enter a valid age.")

            except ValueError:
                print("Please enter a valid number.")

        # Education validation

        while True:

            education = input("Enter your education: ").strip()

            if education:
                break

            print("Education cannot be empty.")

        # Create Student object

        self.student = Student(name, age, education)

    def main(self):

        print("=" * 50)
        print("     SMART CAREER & SKILL ASSESSMENT SYSTEM")
        print("=" * 50)

        # Get student information

        self.get_student_details()

        # Skill assessment

        skills = self.assessment.take_assessment()

        # Career analysis

        self.analyzer = CareerAnalyzer(skills)

        recommended_careers = self.analyzer.assess_career()

        # Display result

        print("\n========================================")
        print("             ASSESSMENT RESULT")
        print("========================================")

        self.student.display_student()

        print("\nRecommended Career:")

        for career in recommended_careers:
            print("✓", career)

        # Display career scores

        print("\n========== CAREER SCORES ==========")

        for career_name, score in self.analyzer.career_scores.items():
            print(career_name, ":", score)

        # Skill analysis

        self.analyzer.show_skill_analysis()

        # Career roadmap

        if len(recommended_careers) == 1:

            self.roadmap.show_roadmap(recommended_careers[0])

        else:

            print("\n========== CAREER ROADMAP ==========")
            print("Multiple careers have the same highest score.")
            print("Explore the roadmaps of these careers.")

        print("\n========================================")
        print("       THANK YOU FOR USING THE SYSTEM")
        print("========================================")


# ==========================================
# CREATE OBJECT
# ==========================================

app = CareerApp()

# START APPLICATION

app.main()
```
